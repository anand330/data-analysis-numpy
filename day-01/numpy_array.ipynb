What is a NumPy Array?
- A NumPy array is a data structure used to store multiple values in a single object.
- NumPy arrays are mainly used to store and process numerical data.
- We create a NumPy array using the np.array() function.
- np.array() is used to create a NumPy array.
- It can convert a Python list into a NumPy array.
- Syntax: np.array(list)


```python
arr=np.array([111,222,333,444,555,666])
print(arr)

list=[11,22,33,44,55,66]
arr2=np.array(list)
print(arr2)
```

```python
# 1. np.array() creates a NumPy array.
arr=np.array([1111,2222,3333,4444])
print(arr)

# 2. It can convert a Python list into a NumPy array.
list=[100,200,300,400]
arr=np.array(list)
print(arr)

```

# 2D Arra

 1. A 2D array has rows and columns.

 
 2. It looks like a table or matrix.
 
 3. It is commonly used to represent tabular numerical data.
 
 4. Each row can contain multiple values.

```python
arr_2d = np.array([[1,2,3],[4,5,6]])
print(arr_2d)

arr2_2d= np.array([[11,22,33],[44,55,66],[77,88,99]])
print(arr2_2d)

arr2_2d.ndim
# ndim       → tells the number of dimensions
```
