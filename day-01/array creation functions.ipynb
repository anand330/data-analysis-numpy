# The important array creation functions are:
 1. np.zeros()
 2. np.ones()
 3. np.arange()
 4. np.linspace()
 5. np.random.randint()
 6. np.random.rand()

# 1. np.zeros()
 1. np.zeros() creates an array filled with zeros.

 2. It can create both 1D and 2D arrays.

 3. np.zeros(5) creates 5 zeros.

 4. np.zeros((2, 3)) creates 2 rows and 3 columns.

 5. It is useful when we need an empty array initialized with zero values.


```python
import numpy as np
arr=np.zeros(10)
print(arr)

arr=np.zeros((10,20))
print(arr)

arr=np.zeros((5,4))
print(arr)
```

# 2. np.ones()
 1. np.ones() creates an array filled with ones.

 2. np.ones(5) creates 5 ones.

 3. np.ones((2, 3)) creates 2 rows and 3 columns.

 4. It is useful when we need an array initialized with one values.



```python
arr=np.ones(10)
print(arr)

arr=np.ones((5,5))
print(arr)

arr=np.ones((10,10))
print(arr)
```

# 3. np.arange()

 1. np.arange() creates a sequence of numbers.

 2. It can use start, stop and step values.

 3. The stop value is not included.

 4. Syntax:   np.arange(start, stop, step)

 5. It is useful for creating numerical sequences.


```python
arr=np.arange(0,11,2)
print(arr)

arr=np.arange(5,51,5)
print(arr)

arr=np.arange(10,101,10)
print(arr)

arr=np.arange(8,81,8)
print(arr)

arr=np.arange(12,121,12)
print(arr)
```

# 4. np.linspace()
 1. np.linspace() creates evenly spaced values.

 2. It uses start, stop and number of values.

 3. The stop value is included by default.

 4. Syntax:   np.linspace(start, stop, num)

 5. It is useful when we need a fixed number of evenly spaced values.


```python
arr= np.linspace(1,20,100)
print(arr)

arr=np.linspace(10,20,50)
print(arr)
```

```python
arr=np.linspace(100,10,5)
print(arr)
```

# 5. np.random.randint()
 1. It generates random integer values.

 2. The low value is included.

 3. The high value is not included.

 4. size tells us how many values or what shape of array we want.

5. The output can be different each time because the values are random.

6. Syntax: np.random.randint(low, high, size)



```python
arr=np.random.randint(1,100,20)
print(arr)

arr= np.random.randint(1,50,10)
print(arr)

arr= np.random.randint(10,20,5)
print(arr)


```

```python
arr=np.random.randint(1,10,(5,5))
print(arr)
```

# 6. np.random.rand()
 1. It generates random decimal values.

 2. Values are between 0 and 1.

 3. 0 can be included, but 1 is not included.

 4. We can create 1D or 2D arrays using this function.

 5. The output can be different each time because the values are random.


```python
arr=np.random.rand(10)
print(arr)
```

```python
arr=np.random.rand(2,5)
print(arr)
```

 np.zeros()        → Creates an array of zeros.

 np.ones()         → Creates an array of ones.

 np.arange()       → Creates a sequence with a step.

 np.linspace()     → Creates evenly spaced values.

 np.random.randint() → Creates random integers.

 np.random.rand()  → Creates random decimal values between 0 and 1.

```python
arr=np.zeros(10)
print(arr)
arr=np.zeros((2,4))
print(arr)

arr=np.ones(10)
print(arr)
arr=np.ones((2,4))
print(arr)

arr=np.linspace(1,10,20)
print(arr)
arr=np.linspace(1,20,5)
print(arr)
arr=np.linspace(0,20,10)
print(arr)
arr=np.linspace(10,20,2)
print(arr)


arr=np.arange(3,31,3)
print(arr)
arr=np.arange(10,101,5)
print(arr)

arr=np.random.randint(1,20,10)
print(arr)
arr=np.random.randint(10,20,10)
print(arr)


arr=np.random.rand(10)
print(arr)
arr=np.random.rand(2,5)
print(arr)
```

np.zeros()          → 0 values

np.ones()           → 1 values

np.arange()         → Sequence + Step

np.linspace()       → Evenly spaced values

np.random.randint() → Random integers

np.random.rand()    → Random decimals (0 to 1)

```python
# Q1. Create an array of 10 zeros.

arr=np.zeros(10)
print(arr)
# Q2. Create an array of 8 ones.

arr=np.ones(8)
print(arr)

# Q3. Create numbers from 1 to 20 using np.arange().

arr=np.arange(1,21,1)
print(arr)
# Q4. Create even numbers from 2 to 20 using np.arange().

arr=np.arange(2,21,2)
print(arr)
# Q5. Create 5 evenly spaced values between 0 and 100
#     using np.linspace().

arr=np.linspace(0,100,5)
print(arr)
# Q6. Generate 10 random integers between 1 and 50.

arr=np.random.randint(1,50,10)
print(arr)

# Q7. Generate a 2 × 3 array of random integers
#     between 10 and 100.

arr= np.random.randint(10,100,(2,3))
print(arr)
                       
# Q8. Generate 5 random decimal values between 0 and 1.

arr=np.random.rand(5)
print(arr)
# Q9. Create a 3 × 4 array filled with zeros.

arr=np.zeros((3,4))
print(arr)
# Q10. Create a 2 × 5 array filled with ones.
arr= np.ones((2,5))
print(arr)
```
